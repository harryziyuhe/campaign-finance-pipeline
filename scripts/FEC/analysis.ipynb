```python
cutoff = 1
df_contrib_stats, df_party_stats, df_pac_race_firm = pd.DataFrame(), pd.DataFrame(), pd.DataFrame()
for i in range(2004, 2026, 2):
    _, election_hedge, _ = get_hedge(i)
    cpac_contrib = race_contributions(i)
    candidates = get_candidates(i)
    house_contrib_stats, house_party_stats, pac_race_firm = house_races(cpac_contrib, candidates, election_hedge)
    house_contrib_stats["YEAR"] = i
    house_party_stats["YEAR"] = i
    house_contrib_stats.to_csv(f"../data/summary/house_contribution_{i}.csv", index=False)
    house_party_stats.to_csv(f"../data/summary/house_contribution_party_{i}.csv", index=False)
    pac_race_firm.to_csv(f"../data/summary/house_compete_contribution_pac_{i}.csv", index=False)
    if len(df_contrib_stats) == 0:
        df_contrib_stats = house_contrib_stats
        df_party_stats = house_party_stats
        df_pac_race_firm = pac_race_firm
    else:
        df_contrib_stats = pd.concat([df_contrib_stats, house_contrib_stats])
        df_party_stats = pd.concat([df_party_stats, house_party_stats])
        df_pac_race_firm = pd.concat([df_pac_race_firm, pac_race_firm])
```

```python
df_contrib_stats.to_csv("../data/summary/house_contribution.csv", index=False)
df_party_stats.to_csv("../data/summary/house_contribution_party.csv", index=False)
df_pac_race_firm.to_csv("../data/summary/house_compete_contribution_pac.csv", index=False)
```
