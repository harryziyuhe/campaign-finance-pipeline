```python
import os
import json 
import pandas as pd

CONFIG_PATH = "S:/campaignfinance/config"
os.environ["LD_LIB_CONFIG_PATH"] = CONFIG_PATH
DATA_PATH = "S:/campaignfinance/data/LSEG/"

import lseg.data as ld
ld.open_session()
```

```python
LSEG_PATH = "S:/campaignfinance/scripts/LSEG/"
def collapse(x):
    x = [v for v in x if pd.notnull(v) and v != '']
    return x if x else None
COLLAPSE_COLS = ["TRBC_ID", "TRBC_Econ_Sector", "TRBC_Business_Sector", "TRBC_Industry_Group",
                 "TRBC_Industry", "TRBC_Activity", "NAICS_ID", "NAICS_Sector", "NAICS_Subsector",
                 "NAICS_Industry_Group", "NAICS_International_Industry", "NAICS_National_Industry"]



with open(f"{LSEG_PATH}data_fields.json", "r") as f:
    data_fields = json.load(f)
    fields = list(data_fields.keys())
    col_names = list(data_fields.values())

agg_dict = {
    col: collapse if col in COLLAPSE_COLS else "first" 
    for col in col_names
}
            
```

```python
df = ld.get_data(universe = "COIN.O", fields = fields)
```

```python
df.columns = ["instrument"] + col_names
```

```python
df.groupby("instrument", as_index=False).agg(agg_dict).reset_index().to_csv(f"{DATA_PATH}Addon_firms.csv", index=False)
```
